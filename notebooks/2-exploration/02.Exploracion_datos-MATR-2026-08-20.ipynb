## 📚 Import  libraries

```python
# base libraries for data science
from pathlib import Path

import pandas as pd
import pyarrow as pa
```

## 💾 Load data

```python
# data directory path
DATA_DIR = Path.cwd().resolve().parents[0] / "Admisiones-predict-project" / "data"

print(f"Data directory path: {DATA_DIR}")

admisiones_df = pd.read_csv(DATA_DIR / "01_raw/Admission_Predict.csv")
```

## Convertir tipos de datos

```python
cols_integer = ["GRE Score", "TOEFL Score", "University Rating"]
admisiones_df[cols_integer] = admisiones_df[cols_integer].astype("Int64")
```

```python
admisiones_df["Research"] = (
    pd.to_numeric(admisiones_df["Research"], errors="coerce")
    .map({0.0: False, 1.0: True})
    .astype("boolean")
)
```

```python
admisiones_df["Research"].value_counts(dropna=False)
```

```python
admisiones_df.info()
```

```python
schema = pa.Table.from_pandas(admisiones_df).schema
```

## 💾 Save dataframe with data types

```python
processed_path = DATA_DIR / "02_intermediate/admisiones_type_fixed.parquet"
admisiones_df.to_parquet(
    processed_path,
    index=False,
    schema=schema,
)
pd.read_parquet(processed_path).info()
```

## 📊 Analysis of Results and Conclusions 

### Resultados y conclusiones

El conjunto de datos contiene 623 registros y 8 variables. La variable objetivo,
`Chance of Admit`, no presenta valores faltantes y tiene un promedio de 0.73,
con valores entre 0.34 y 0.97.

Se identificaron valores faltantes en las variables predictoras. `Research`
presenta la mayor proporción de datos faltantes (29 registros, 4.65 %), seguida
por `TOEFL Score` (19 registros, 3.05 %) y `SOP` (16 registros, 2.57 %).
Las demás variables presentan menos del 2 % de valores faltantes. Los valores
faltantes se conservaron como valores nulos, evitando reemplazarlos por cero o
por estadísticas sin una justificación basada en el contexto o en el modelo.

Se corrigieron los tipos de datos para representar adecuadamente el esquema:
`GRE Score`, `TOEFL Score` y `University Rating` se almacenaron como enteros
anulables; `Research` como booleano anulable; y las variables `SOP`, `LOR`,
`CGPA` y `Chance of Admit` se conservaron como numéricas decimales.

Finalmente, el dataset procesado se almacena en formato Parquet, que preserva
el esquema y los valores faltantes. Para la etapa de modelado se deberá definir
una estrategia de imputación usando únicamente los datos de entrenamiento,
con el fin de evitar fuga de información.
