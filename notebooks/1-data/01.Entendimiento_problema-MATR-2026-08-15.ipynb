## 1. ¿Cuál es el objetivo del problema?

Estimar la variable continua `Chance of Admit `, interpretada como la probabilidad o puntaje de posibilidad de admision a un programa de posgrado, a partir del perfil academico disponible.

## 2. ¿Cómo se usará su solución?

## 4. ¿Cómo se debe enmarcar este problema? (supervisado / no supervisado, en línea / fuera de línea, etc.)

Es un problema de aprendizaje supervisado de regresion: la variable objetivo es numerica y las demas columnas son posibles predictores. La primera version debe entrenarse y evaluarse fuera de linea sobre un conjunto historico separado. ## 5. ¿Cómo se debe medir el desempeño o rendimiento de la solución?

Primera intuicion: usar MAE como metrica principal sobre datos no vistos, porque comunica el error medio en la misma escala de `Chance of Admit ` (0 a 1 en el archivo local). Reportar tambien RMSE para penalizar errores grandes y R2 como referencia comparativa. Antes de escoger una metrica final se debe confirmar si la etiqueta es una probabilidad, una calificacion subjetiva o un resultado observado.

## 6. ¿La medida de desempeño está alineada con el objetivo del problema?

MAE esta alineada con el objetivo porque mide, en promedio, cuan lejos esta la probabilidad estimada de la referencia en la misma escala de 0 a 1. Por ejemplo, un MAE de 0.08 equivale a un error promedio de 8 puntos porcentuales.

## 7. ¿Cuál sería el desempeño mínimo necesario para alcanzar el objetivo?

Para este trabajo yo diria que un MAE menor a 0.3 como referencial inicial y porque ya seria un modelo que es mejor que tirar una moneda.
